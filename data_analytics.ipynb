```python
from scipy.stats import f_oneway

# Create groups for each irrigation method
groups = [
    group["Yield_Tonnes_Ha"].dropna()
    for name, group in df.groupby("Irrigation_Method")
]

# Perform one-way ANOVA
f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: The difference in yield between irrigation methods is statistically significant.")
else:
    print("Result: The difference in yield between irrigation methods is not statistically significant.")
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Remove rows where yield or irrigation method is missing
tukey_df = df[["Yield_Tonnes_Ha", "Irrigation_Method"]].dropna()

# Perform Tukey HSD test
tukey_result = pairwise_tukeyhsd(
    endog=tukey_df["Yield_Tonnes_Ha"],
    groups=tukey_df["Irrigation_Method"],
    alpha=0.05
)

print(tukey_result)
```

```python
from scipy.stats import f_oneway

# Create yield groups for each season
season_groups = [
    group["Yield_Tonnes_Ha"].dropna()
    for name, group in df.groupby("Season")
]

# One-way ANOVA
f_stat_season, p_value_season = f_oneway(*season_groups)

print("Season ANOVA")
print("F-statistic:", f_stat_season)
print("P-value:", p_value_season)

if p_value_season < 0.05:
    print("Result: Yield differences between seasons are statistically significant.")
else:
    print("Result: Yield differences between seasons are not statistically significant.")
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

season_tukey_df = df[["Yield_Tonnes_Ha", "Season"]].dropna()

season_tukey = pairwise_tukeyhsd(
    endog=season_tukey_df["Yield_Tonnes_Ha"],
    groups=season_tukey_df["Season"],
    alpha=0.05
)

print(season_tukey)
```
